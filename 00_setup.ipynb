```python
%sql
-- Create a separate catalog named "finops" to store tables related to this project 
CREATE CATALOG IF NOT EXISTS finops
MANAGED LOCATION 'abfss://finops@harshifinops01.dfs.core.windows.net/managed';
```

```python
%sql
-- Create a SCHEMA = A high-level grouping or namespace, lives inside a Catalog
-- Create a VOLUME = A place to hold files, lives inside a Schema 
CREATE SCHEMA IF NOT EXISTS finops.raw;
CREATE VOLUME IF NOT EXISTS finops.raw.real_exports;
```

```python
%sql
-- Schemas the pipeline writes into (the pipeline creates tables, not schemas)
CREATE SCHEMA IF NOT EXISTS finops.bronze;
CREATE SCHEMA IF NOT EXISTS finops.silver;
CREATE SCHEMA IF NOT EXISTS finops.gold;
```
